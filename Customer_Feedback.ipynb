Step 1: Import & Prepare Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("/content/onlinefoods.csv")
```

```python
# View feedback value counts
feedback_counts = df['Feedback'].value_counts()
print("Feedback Distribution:")
print(feedback_counts)
```

Step 2: Visualize Feedback Sentiment

```python
# Set visualization style
sns.set(style="whitegrid")
```

```python
# Bar chart for sentiment
plt.figure(figsize=(6, 4))
sns.countplot(x='Feedback', data=df, palette='RdYlGn')
plt.title("Customer Feedback Sentiment Distribution")
plt.xlabel("Feedback")
plt.ylabel("Number of Responses")
plt.tight_layout()
plt.show()
```

 Step 3: Quantify Sentiment with Metrics (CSAT & NPS)


```python
# Total responses
total_responses = len(df)
negative = df['Feedback'].str.strip().str.lower().value_counts().get('negative', 0)
positive = df['Feedback'].str.strip().str.lower().value_counts().get('positive', 0)
```

```python
# CSAT: % of satisfied customers
csat = round((positive / total_responses) * 100, 2)
```

```python
# NPS Approximation
# Assuming Positive = Promoters, Negative = Detractors
nps = round(((positive - negative) / total_responses) * 100, 2)

print(f"Customer Satisfaction Score (CSAT): {csat}%")
print(f"Net Promoter Score (NPS Approx.): {nps}")
```

Step 4: Business Analysis Summary


```python
# Automated insight generation (for report)
summary = f"""
📊 FEEDBACK ANALYSIS SUMMARY:

• Total Responses: {total_responses}
• Positive Feedback: {positive}
• Negative Feedback: {negative}

📈 METRICS:
• Customer Satisfaction Score (CSAT): {csat}%
• Net Promoter Score (NPS - Simulated): {nps}

🔍 INTERPRETATION:
• The business maintains a strong satisfaction level among customers, with {csat}% expressing positive sentiment.
• A high NPS of {nps} indicates a strong likelihood of customer retention and referrals.
• However, {negative} customers expressed dissatisfaction — these cases should be analyzed to identify root causes and improve service quality.

🎯 RECOMMENDATIONS:
• Focus on maintaining strengths (timely delivery, quality meals).
• Collect detailed feedback to understand issues causing negative responses.
• Implement loyalty and referral programs to leverage high promoter sentiment.
"""
print(summary)
```
