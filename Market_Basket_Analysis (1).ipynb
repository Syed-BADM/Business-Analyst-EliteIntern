```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv(r"/content/Assignment-1_Data.csv", delimiter=";")
```

```python
df["Price"] = df["Price"].str.replace(",", ".").astype(float)
df.dropna(subset=["Itemname", "BillNo"], inplace=True)
df["BillNo"] = df["BillNo"].astype(str)
```

```python
transactions = df.groupby("BillNo")["Itemname"].apply(list).tolist()
```

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)
basket = pd.DataFrame(te_array, columns=te.columns_)
```

```python
frequent_itemsets = apriori(basket, min_support=0.02, use_colnames=True)
```

```python
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
```

```python
print("Frequent Itemsets:\n", frequent_itemsets.head())
print("\nAssociation Rules:\n", rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
```
